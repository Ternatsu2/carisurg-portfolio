For Day 1, I loaded the reduced emergency triage dataset and cleaned the "Gender" column only. I kept the rest of the dataset unchanged so the notebook stays focused on the assignment.


```python
import sys
from pathlib import Path

import pandas as pd

print(f"Python version: {sys.version}")
print(f"pandas version: {pd.__version__}")
```

## Load the dataset

I loaded the CSV into pandas and checked the row and column count.


```python
candidate_paths = [
    Path("EmergencyTriageDataset_Reduced_Dirty.csv"),
    Path("../data/EmergencyTriageDataset_Reduced_Dirty.csv"),
    Path("../source_materials/week0/EmergencyTriageDataset_Reduced_Dirty.csv"),
    Path("../../../source_materials/week0/EmergencyTriageDataset_Reduced_Dirty.csv"),
    Path("/content/drive/MyDrive/ColabNotebooks/CariSurg_Triage_Test/EmergencyTriageDataset_Reduced_Dirty.csv"),
]

FILE_PATH = next((path for path in candidate_paths if path.exists()), None)
if FILE_PATH is None:
    raise FileNotFoundError("Upload the Week 0 CSV or update FILE_PATH.")

df_raw = pd.read_csv(FILE_PATH)
print(f"Loaded {df_raw.shape[0]} rows and {df_raw.shape[1]} columns from {FILE_PATH}")
df_raw.head()
```

## Check the original Gender values

Before changing the column, I checked which values appeared in "Gender".


```python
print("Original Gender values:")
print(df_raw["Gender"].value_counts(dropna=False))
```

## Clean the Gender column

The same values were written in a few different ways. I mapped "Male", "MALE", and "1" to "1". I mapped "Female", "FEMALE", and "0" to "0".


```python
gender_map = {
    "male": 1,
    "female": 0,
    "1": 1,
    "0": 0,
}

df = df_raw.copy()
gender_text = df["Gender"].astype("string").str.strip().str.lower()
df["Gender_Clean"] = gender_text.map(gender_map).astype("Int64")

unmatched = df.loc[df["Gender_Clean"].isna(), "Gender"].dropna().unique()
print(f"Unmatched non-empty Gender values: {list(unmatched)}")
print("\nCleaned Gender values:")
print(df["Gender_Clean"].value_counts(dropna=False).sort_index())
```

## Replace the dirty column

After checking that the mapping worked, I replaced the original column with the cleaned version.


```python
df = df.drop(columns=["Gender"]).rename(columns={"Gender_Clean": "Gender"})

print(df["Gender"].value_counts(dropna=False).sort_index())
print(f"Gender dtype: {df['Gender'].dtype}")
df.head()
```

## Short note

The cleaned column has 1,176 records coded as "1" and 1,029 records coded as "0". No values were left unmatched.
